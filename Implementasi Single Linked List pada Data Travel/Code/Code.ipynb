```python
class Node:
    def __init__(self, Traveler_age, Traveler_gender, Travel_nasionality, Accomodation_type, Accomodation_cost, Transportation_type, Transportation_cost): # Inisialisasi atribut-atribut dari objek Node
        self.Traveler_age = Traveler_age  # Usia traveler
        self.Traveler_gender = Traveler_gender  # Jenis kelamin traveler
        self.Travel_nasionality = Travel_nasionality  # Kebangsaan traveler
        self.Accomodation_type = Accomodation_type  # Tipe akomodasi yang digunakan oleh traveler
        self.Accomodation_cost = Accomodation_cost  # Biaya akomodasi traveler
        self.Transportation_type = Transportation_type  # Tipe transportasi yang digunakan oleh traveler
        self.Transportation_cost = Transportation_cost  # Biaya transportasi traveler
        self.next = None  # Pointer ke Node selanjutnya dalam linked list


class LinkedList: #kelas linked list
    def __init__(self):
        self.head = None #head belum ada

    def sort(self):
      if self.head is None:  # Jika linked list kosong, tidak perlu dilakukan sorting
        return

      last = self.head  # Inisialisasi variabel last untuk menunjuk ke node terakhir
      while (last.next):  # Mencari node terakhir dalam linked list
        last = last.next

      current = self.head  # Inisialisasi variabel current untuk iterasi pertama pada linked list
      while current.next:  # Iterasi pada linked list
        if current.data > current.next.data:  # Membandingkan data node saat ini dengan data node berikutnya
            # Jika data node saat ini lebih besar dari data node berikutnya, tukar posisinya
            current.data, current.next.data = current.next.data, current.data
        current = current.next  # Pindah ke node berikutnya


    def add_at_beginning(self, Traveler_age, Traveler_gender, Travel_nasionality, Accomodation_type, Accomodation_cost, Transportation_type, Transportation_cost):
      new_node = Node(Traveler_age, Traveler_gender, Travel_nasionality, Accomodation_type, Accomodation_cost, Transportation_type, Transportation_cost) # Membuat node baru dengan data yang diberikan
      new_node.next = self.head # Mengatur pointer node baru agar menunjuk ke node yang sebelumnya menjadi kepala linked list

    # Mengatur node baru sebagai kepala linked list
      self.head = new_node  # Memberikan node baru sebagai kepala linked list


    def add_at_end(self, Traveler_age, Traveler_gender, Travel_nasionality, Accomodation_type, Accomodation_cost, Transportation_type, Transportation_cost):
      new_node = Node(Traveler_age, Traveler_gender, Travel_nasionality, Accomodation_type, Accomodation_cost, Transportation_type, Transportation_cost) # Membuat node baru dengan data yang diberikan
      if self.head is None:  # Memeriksa apakah linked list kosong
        self.head = new_node  # Jika linked list kosong, node baru menjadi node pertama
        return
      last = self.head  # Inisialisasi variabel last untuk menunjuk ke node terakhir
      while (last.next):  # Mencari node terakhir dalam linked list
        last = last.next
      last.next = new_node  # Menambahkan node baru setelah node terakhir


    def insert_at_beginning(self, Traveler_age, Traveler_gender, Travel_nasionality, Accomodation_type, Accomodation_cost, Transportation_type, Transportation_cost):
      new_node = Node(Traveler_age, Traveler_gender, Travel_nasionality, Accomodation_type, Accomodation_cost, Transportation_type, Transportation_cost) # Membuat node baru dengan data yang diberikan

    # Menyisipkan node baru ke awal linked list
      new_node.next = self.head  # Mengatur pointer node baru agar menunjuk ke node yang sebelumnya menjadi kepala linked list
      self.head = new_node  # Mengatur node baru sebagai kepala linked list


    def insert_at_end(self, Traveler_age, Traveler_gender, Travel_nasionality, Accomodation_type, Accomodation_cost, Transportation_type, Transportation_cost):
      new_node = Node(Traveler_age, Traveler_gender, Travel_nasionality, Accomodation_type, Accomodation_cost, Transportation_type, Transportation_cost) # Membuat node baru dengan data yang diberikan
      if not self.head:  # Memeriksa apakah linked list kosong
        self.head = new_node  # Jika linked list kosong, node baru menjadi node pertama
        return

    # Jika linked list tidak kosong, tambahkan node baru di belakang node terakhir
      current = self.head
      while current.next:
        current = current.next
      current.next = new_node  # Menambahkan node baru setelah node terakhir


    def delete_from_beginning(self):
        if not self.head: #memeriksa linked list lagi
            print("List is empty. Nothing to delete.")
        else:
            self.head = self.head.next #mengganti head linked list ke node selanjutnya

    def delete_from_end(self):
      if not self.head:  # Memeriksa apakah linked list kosong
        print("List is empty. Nothing to delete.")
      elif not self.head.next:  # Memeriksa apakah linked list hanya memiliki satu node
        self.head = None  # Jika hanya ada satu node, maka hapus node tersebut
      else:
        current = self.head
        while current.next.next:  # Mencari node kedua terakhir dalam linked list
            current = current.next
        current.next = None  # Menghapus node terakhir dengan mengubah pointer node kedua terakhir menjadi None


    def print_data(self):
      current = self.head
      if not current:
        print("List is empty.")  # Memeriksa apakah linked list kosong
      else:
        print("Data in the list Traveler:")  # Mencetak data dari linked list
        while current:
            # Mencetak data dari setiap node dalam linked list
            print(f"Age: {current.Traveler_age}, Gender: {current.Traveler_gender}, Nationality: {current.Travel_nasionality}, Accommodation Type: {current.Accomodation_type}, Accommodation Cost: {current.Accomodation_cost}, Transportation Type: {current.Transportation_type}, Transportation Cost: {current.Transportation_cost}")
            current = current.next


    def display(self):
      elements = []  # Inisialisasi list untuk menyimpan elemen
      current_node = self.head  # Mulai dari kepala linked list

    # Mengambil setiap elemen dari linked list
      while current_node:
        elements.append(str(current_node.data))  # Menambahkan data dari setiap node ke dalam list
        current_node = current_node.next  # Pindah ke node berikutnya

    # Mencetak elemen-elemen dari linked list yang telah dikumpulkan
      print(" -> ".join(elements))  # Menggabungkan elemen-elemen dalam list sebagai output dengan pemisah " -> "


    def reverse(self):
      prev_node = None  # Node sebelumnya diinisialisasi sebagai None
      current_node = self.head  # Node saat ini diinisialisasi sebagai kepala linked list

    # Melakukan proses pembalikan linked list
      while current_node is not None:
        next_node = current_node.next  # Menyimpan referensi ke node selanjutnya
        current_node.next = prev_node  # Merubah pointer ke node sebelumnya untuk melakukan pembalikan
        prev_node = current_node  # Pindah ke node berikutnya
        current_node = next_node  # Pindah ke node selanjutnya

      self.head = prev_node  # Mengubah kepala linked list menjadi node terakhir setelah pembalikan
```

```python
def append_to_list(self, data):
    if not self.head:  # Memeriksa apakah linked list kosong
        self.head = Node(data)  # Jika kosong, node baru akan menjadi node pertama
    else:
        current = self.head
        while current.next:  # Mencari node terakhir dalam linked list
            current = current.next
        current.next = Node(data)  # Menambahkan node baru setelah node terakhir
```
